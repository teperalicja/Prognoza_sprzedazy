```python
import pandas as pd

sales = pd.read_csv(r"C:\Users\Alicja\Desktop\kurs-datascience\zadanie_rekrutacyjne_junior\zadanie_rekrutacyjne_junior\data\sales data-set.csv")
features = pd.read_csv(r"C:\Users\Alicja\Desktop\kurs-datascience\zadanie_rekrutacyjne_junior\zadanie_rekrutacyjne_junior\data\Features data set.csv")
stores = pd.read_csv(r"C:\Users\Alicja\Desktop\kurs-datascience\zadanie_rekrutacyjne_junior\zadanie_rekrutacyjne_junior\data\stores data-set.csv")

print(sales.shape, features.shape, stores.shape)
```

Łączenie tabel:

```python
print("Sales kolumny:", list(sales.columns))
print("Features kolumny:", list(features.columns))
print("Stores kolumny:", list(stores.columns))

sales["Date"] = pd.to_datetime(sales["Date"], format="%d/%m/%Y")
features["Date"] = pd.to_datetime(features["Date"], format="%d/%m/%Y")

df = ( sales
.merge(stores, on='Store', how="left")
.merge(features.drop(columns='IsHoliday'), on=["Store","Date"], how="left"))

print("\nKolumny nowej tabeli:", list(df.columns))
```

```python
print("Kolumny zawierające brakujące informacji:\n", df.isna().sum())
```

Zamiana wartości "NA" na 0.

```python
markdown = ["MarkDown1", "MarkDown2", "MarkDown3", "MarkDown4", "MarkDown5"]

df[markdown] = df[markdown].fillna(0)
df[markdown].head()
```

```python
print(df[markdown].isna().sum())
```

Oznaczenie tygodni, w których była promocja.

```python
df["IsPromo"] = (df[markdown].sum(axis=1) > 0).astype(int)
df[markdown].shape
```
